```python
gender_attrition = data.groupby('Gender')['Attrition_Flag'].value_counts(normalize=True).unstack()
gender_attrition['Attrited Customer Percentage'] = gender_attrition['Attrited Customer'] * 100
gender_attrition= gender_attrition.sort_values(by='Attrited Customer Percentage', ascending=False)

#creating a bar plot to visualize the attrition
plt.figure(figsize=(8,6))
gender_attrition['Attrited Customer Percentage'].plot(kind='bar')
plt.title('Attrition Rate by Gender')
plt.xlabel('Gender')
plt.ylabel('Attrited Customer Percentage')
plt.xticks(rotation=0)
plt.show()
```

This code performs **bivariate analysis** to calculate and visualize the **customer attrition rate by gender**.

* `groupby('Gender')` → Groups customers based on gender.
* `value_counts(normalize=True)` → Calculates the proportion of each attrition category.
* `unstack()` → Converts the results into separate columns.
* `* 100` → Converts the attrited customer proportion into a percentage.
* `sort_values()` → Arranges genders from highest to lowest attrition percentage.
* `plot(kind='bar')` → Creates a bar chart to compare attrition rates.
* `plt.xticks(rotation=0)` → Keeps gender labels horizontal.

**Purpose:** To compare the **percentage of attrited customers between genders** and identify which gender has a higher attrition rate.

# **DATA PREPROCESSING**


Data preprocessing is the process of preparing the dataset for machine learning analysis.
Missing values and duplicate records are identified and handled properly.
Categorical variables are converted into numerical values using encoding techniques.

These steps help improve the accuracy and performance of the machine learning model.

```python
data = data.drop('CLIENTNUM', axis=1)
```


The CLIENTNUM column is removed from the dataset because it is a unique customer identification number.
It does not provide useful information for predicting customer attrition.

```python
data['Attrition_Flag'].value_counts()
```

The value_counts() function is used to count the number of customers in each category of the Attrition_Flag column.
It helps to understand how many customers are Existing and how many are Attrited.
This is useful for checking the distribution of the target variable before building the machine learning model.

```python
Attrition_mapping = {'Existing Customer': 0, 'Attrited Customer': 1}
data['Attrition_Flag'] = data['Attrition_Flag'].map(Attrition_mapping)
```

The Attrition_Flag column contains categorical values, so they are converted into numerical values.
Existing Customer is mapped to 0, and Attrited Customer is mapped to 1.
This conversion makes the target variable suitable for machine learning algorithms.

```python
data['Education_Level'].unique()
```


The unique() function is used to find all the different categories present in the Education_Level column.
It helps identify the unique education levels available in the dataset.

```python
data.columns
```

```python
Education_maping = {'Uneducated':1,'High School':2, 'College':3, 'Graduate':4, 'Post-Graduate':5 ,'Doctorate':6, 'Unknown':0}
data['Education_Level'] = data['Education_Level'].map(Education_maping)
```

The Education_Level column contains categorical values, so they are converted into numerical values.
Each education level is assigned a specific number from 1 to 6.
The Unknown category is assigned 0.
This encoding makes the data suitable for machine learning algorithms.

```python
data = pd.get_dummies(data, columns=['Marital_Status', 'Card_Category'], drop_first=True, dtype=int)
```

The Marital_Status and Card_Category columns are categorical variables.

```python
model_eval_results = pd.DataFrame(columns=['Model', 'Train_Accuracy', 'Train_Precision', 'Train_Recall', 'Train_F1','Test_Accuracy', 'Test_Precision', 'Test_Recall', 'Test_F1'])
```

## **Model 1**

```python
dt_model = DecisionTreeClassifier(random_state=42)
```

```python
dt_model.fit(x_train, y_train)
```

```python
y_train_pred_dt = dt_model.predict(x_train)
y_test_pred_dt = dt_model.predict(x_test)
```

```python
train_accuracy_dt = accuracy_score(y_train, y_train_pred_dt)
train_precision_dt = precision_score(y_train, y_train_pred_dt)
train_recall_dt = recall_score(y_train, y_train_pred_dt)
train_f_dt = f1_score(y_train, y_train_pred_dt)
```

```python
test_accuracy_dt = accuracy_score(y_test, y_test_pred_dt)
test_precision_dt = precision_score(y_test, y_test_pred_dt)
test_recall_dt = recall_score(y_test, y_test_pred_dt )
test_f1_dt = f1_score(y_test, y_test_pred_dt)
```

```python
model_eval_results.loc[len(model_eval_results)] = ['Decision Tree', train_accuracy_dt, train_precision_dt, train_recall_dt, train_f1_dt, test_accuracy_dt, test_precision_dt, test_recall_dt, test_f1_dt]
```
